```python
# Creating a backup of the original dataset

df_raw = df.copy()

print("Original dataset backup created successfully!")
```

```python
# Removing exact duplicate records

df_clean = df_raw.drop_duplicates().copy()

print("Original Records:", len(df_raw))
print("Records After Removing Duplicates:", len(df_clean))
print("Duplicate Rows Removed:", len(df_raw) - len(df_clean))
```

```python
# Standardizing gender values

df_clean["Gender"] = (
    df_clean["Gender"]
    .str.strip()
    .str.lower()
    .replace({
        "m": "male",
        "f": "female"
    })
    .str.title()
)

print(df_clean["Gender"].value_counts())
```

```python
# Standardizing region names

df_clean["Region"] = (
    df_clean["Region"]
    .str.strip()
    .str.lower()
    .replace({
        "nort": "north",
        "sth": "south"
    })
    .str.title()
)

print(df_clean["Region"].value_counts())
```

```python
# Checking cleaned categorical values

print("Gender Categories:")
print(df_clean["Gender"].unique())

print("\nRegion Categories:")
print(df_clean["Region"].unique())

print("\nRemaining Exact Duplicates:",
      df_clean.duplicated().sum())
```

```python
# Identifying invalid age values

df_clean.loc[
    (df_clean["Age"] < 0) | (df_clean["Age"] > 100),
    "Age"
] = np.nan

print("Invalid ages converted to missing values.")
```

```python
# Handling invalid discount values

df_clean.loc[
    (df_clean["Discount"] < 0) | (df_clean["Discount"] > 0.5),
    "Discount"
] = np.nan

print("Invalid discounts converted to missing values.")
```

```python
# Filling missing values using median

df_clean["Age"] = df_clean["Age"].fillna(
    df_clean["Age"].median()
)

df_clean["Income"] = df_clean["Income"].fillna(
    df_clean["Income"].median()
)

df_clean["Discount"] = df_clean["Discount"].fillna(
    df_clean["Discount"].median()
)

print("Missing numerical values handled successfully!")
```

```python
# Converting date columns into datetime format

df_clean["Order_Date"] = pd.to_datetime(
    df_clean["Order_Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

df_clean["Delivery_Date"] = pd.to_datetime(
    df_clean["Delivery_Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print("Date conversion completed!")
```

```python
# Identifying delivery dates earlier than order dates

invalid_delivery = (
    df_clean["Delivery_Date"] < df_clean["Order_Date"]
)

print("Invalid delivery dates:", invalid_delivery.sum())
```

```python
# Flagging invalid delivery dates

df_clean["Delivery_Date_Flag"] = invalid_delivery

df_clean.loc[
    invalid_delivery,
    "Delivery_Date"
] = pd.NaT

print("Invalid delivery dates flagged successfully!")
```

```python
# Checking remaining missing values

print("Remaining Missing Values:\n")

print(df_clean.isnull().sum())
```

```python
# Checking repeated transaction IDs

repeated_ids = df_clean[
    df_clean["Transaction_ID"].duplicated(keep=False)
]

print("Repeated Transaction ID Records:")
print(repeated_ids.sort_values("Transaction_ID").head(20))

print(
    "Remaining repeated ID records:",
    len(repeated_ids)
)
```

```python
# Saving intermediate cleaned dataset

df_clean.to_csv(
    "../Dataset/retail_sales_intermediate_cleaned.csv",
    index=False
)

print("Intermediate cleaned dataset saved successfully!")
```

```python
# Comparing original and converted delivery dates

print("Original Missing Dates:",
      df_raw["Delivery_Date"].isnull().sum())

print("Converted Missing Dates:",
      df_clean["Delivery_Date"].isnull().sum())

print("\nOriginal Date Examples:")
print(df_raw["Delivery_Date"].head(15).to_string())
```

```python
# Checking dates that failed during conversion

invalid_dates = df_raw[
    df_raw["Delivery_Date"].notna() &
    pd.to_datetime(
        df_raw["Delivery_Date"],
        format="mixed",
        dayfirst=True,
        errors="coerce"
    ).isna()
]

print("Dates that failed conversion:",
      len(invalid_dates))

print(invalid_dates["Delivery_Date"].head(20).to_string())
```

```python
print("Original Missing Dates:", df_raw["Delivery_Date"].isnull().sum())
print("Converted Missing Dates:", df_clean["Delivery_Date"].isnull().sum())

print("\nOriginal Date Examples:")
print(df_raw["Delivery_Date"].head(15).to_string())
```

```python
# Recreate dates using original data
order_dates = pd.to_datetime(
    df_raw.loc[df_clean.index, "Order_Date"],
    errors="coerce"
)

delivery_dates = pd.to_datetime(
    df_raw.loc[df_clean.index, "Delivery_Date"],
    errors="coerce"
)

# Check dates before order date
invalid_check = delivery_dates < order_dates

print("Original Missing Dates:", delivery_dates.isna().sum())
print("Dates Before Order Date:", invalid_check.sum())
print("Total Rows:", len(df_clean))

print("\nSample Date Comparison:")
print(pd.DataFrame({
    "Order_Date": order_dates,
    "Delivery_Date": delivery_dates
}).head(15))
```

```python
# Convert dates using mixed format
order_dates = pd.to_datetime(
    df_raw.loc[df_clean.index, "Order_Date"],
    format="mixed",
    errors="coerce"
)

delivery_dates = pd.to_datetime(
    df_raw.loc[df_clean.index, "Delivery_Date"],
    format="mixed",
    errors="coerce"
)

# Check missing and invalid dates
print("Missing Order Dates:", order_dates.isna().sum())
print("Missing Delivery Dates:", delivery_dates.isna().sum())

invalid_check = (
    delivery_dates.notna() &
    order_dates.notna() &
    (delivery_dates < order_dates)
)

print("Delivery Before Order:", invalid_check.sum())

print("\nInvalid Date Examples:")
print(pd.DataFrame({
    "Order_Date": order_dates,
    "Delivery_Date": delivery_dates
})[invalid_check].head(15))
```

```python
# Apply correct date conversion
df_clean["Order_Date"] = order_dates
df_clean["Delivery_Date"] = delivery_dates

# Identify invalid delivery dates
invalid_delivery = (
    df_clean["Delivery_Date"].notna() &
    (df_clean["Delivery_Date"] < df_clean["Order_Date"])
)

# Keep a record of invalid dates
df_clean["Delivery_Date_Flag"] = invalid_delivery

# Replace invalid delivery dates with NaT
df_clean.loc[invalid_delivery, "Delivery_Date"] = pd.NaT

# Calculate delivery duration
df_clean["Delivery_Days"] = (
    df_clean["Delivery_Date"] - df_clean["Order_Date"]
).dt.days

print("Invalid Delivery Dates:", invalid_delivery.sum())
print("Remaining Missing Delivery Dates:",
      df_clean["Delivery_Date"].isna().sum())

print("\nDelivery Days Summary:")
print(df_clean["Delivery_Days"].describe())
```

```python
df_clean.to_csv(
    "../Dataset/retail_sales_cleaned.csv",
    index=False
)

print("Final Cleaned Dataset Saved Successfully!")
print("Final Dataset Shape:", df_clean.shape)
```

```python
# Check deliveries taking more than 30 days
long_deliveries = df_clean[df_clean["Delivery_Days"] > 30]

print("Deliveries taking more than 30 days:",
      len(long_deliveries))

print("\nLongest Delivery Records:")
print(
    long_deliveries[
        ["Order_Date", "Delivery_Date", "Delivery_Days",
         "Region", "Product_Category"]
    ].sort_values("Delivery_Days", ascending=False).head(10)
)
```

```python
# Basic Business Analysis

print("Total Sales:", df_clean["Sales"].sum())
print("Total Profit:", df_clean["Profit"].sum())
print("Total Transactions:", df_clean["Transaction_ID"].nunique())
print("Average Order Value:", df_clean["Sales"].mean())
print("Total Quantity Sold:", df_clean["Quantity"].sum())

print("\nSales by Region:")
print(
    df_clean.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nSales by Product Category:")
print(
    df_clean.groupby("Product_Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_style("whitegrid")

os.makedirs("../VISUALIZATION", exist_ok=True)

# 1. Region-wise Sales
region_sales = df_clean.groupby("Region")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(9, 5))
sns.barplot(x=region_sales.index, y=region_sales.values)
plt.title("Region-wise Sales Analysis")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.savefig("../VISUALIZATION/region_sales.png")
plt.show()
plt.close()


# 2. Product Category Sales
category_sales = df_clean.groupby("Product_Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(x=category_sales.index, y=category_sales.values)
plt.title("Product Category-wise Sales")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig("../VISUALIZATION/category_sales.png")
plt.show()
plt.close()


# 3. Monthly Sales Trend
monthly_sales = df_clean.groupby(
    df_clean["Order_Date"].dt.to_period("M")
)["Sales"].sum()

plt.figure(figsize=(12, 5))
monthly_sales.index = monthly_sales.index.astype(str)
monthly_sales.plot(marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../VISUALIZATION/monthly_sales.png")
plt.show()
plt.close()


# 4. Category-wise Profit
category_profit = df_clean.groupby("Product_Category")["Profit"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(x=category_profit.index, y=category_profit.values)
plt.title("Category-wise Profit Analysis")
plt.xlabel("Product Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig("../VISUALIZATION/category_profit.png")
plt.show()
plt.close()

print("All 4 charts saved successfully!")
```

```python
# Customer and business insights

transaction_sales = df_clean.groupby("Transaction_ID")["Sales"].sum()

print("Average Order Value:",
      round(transaction_sales.mean(), 2))

print("Profit Margin (%):",
      round(
          (df_clean["Profit"].sum() / df_clean["Sales"].sum()) * 100,
          2
      ))

print("\nCustomer Segment-wise Sales:")
print(
    df_clean.groupby("Customer_Segment")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nPayment Method-wise Transactions:")
print(df_clean["Payment_Method"].value_counts())
```

# Conclusion 

The Retail Customer Sales Exploratory Data Analysis project was conducted to understand sales performance, profitability, customer purchasing patterns, and product category performance.

The dataset contained 7,084 records initially. After removing 61 duplicate rows, 7,023 records remained for analysis.

The analysis produced the following findings:

1. Total sales were approximately ₹7.70 crore.
2. Total profit was approximately ₹2.52 crore.
3. The East region recorded the highest sales among all four regions.
4. Electronics was the highest-selling product category, followed by Furniture.
5. The average sales value per record was approximately ₹10,958.
6. The analysis also identified unusual delivery durations that may require further investigation.

The project provided practical experience in data cleaning, data analysis, grouping, and visualization using Python libraries such as Pandas, NumPy, Matplotlib, and Seaborn.

Overall, this project demonstrates how exploratory data analysis can help businesses understand their sales data and make better data-driven decisions.
